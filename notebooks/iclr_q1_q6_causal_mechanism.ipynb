**Companion to `highnorm_sinks_massive_channels_v5.ipynb`.** v5 establishes the observational
picture: high-norm register tokens, attention sinks and massive activation channels are one
phenomenon organised around a model-specific register direction $v^*$, and in FLUX that direction
is almost a single coordinate. This notebook does not re-establish any of that. It runs the six
experiments that are still open, and it *runs* them: every question below executes, records
pre-declared endpoints, and prints a verdict.

## The causal model under test

$$\text{writer}\;\longrightarrow\;c^*\;\longrightarrow\;v^*\;\longrightarrow\;\text{key geometry}
\;\longrightarrow\;\text{sink routing}\;\longrightarrow\;\text{dissolution}$$

| | Question | Arrow it closes |
|---|---|---|
| **Q1** | If the natural register tokens are removed, where do the sinks go? | key geometry → sink routing |
| **Q2** | What happens downstream after live dominant-channel suppression? | $c^*$ → $v^*$ |
| **Q3** | Does the network regenerate or relocate $v^*$ after we destroy it? | is the state maintained? |
| **Q4** | What selects the sparse token positions that receive the write? | writer → $c^*$ |
| **Q5** | What makes $v^*$ sufficient at a natural register but not elsewhere? | $v^*$ → key geometry |
| **Q6** | What causes the register state to dissolve? | sink routing → dissolution |

## Three rules this notebook enforces in code, not in prose

1. **Targets are frozen.** Every intervention edits token identities chosen from a *clean* run of
   the same prompt and seed. No primary endpoint is ever computed from a set reselected out of a
   treated activation.
2. **Comparisons are paired.** One experimental unit is one `(prompt, seed)`. The clean pass runs
   first; each condition then replays the identical generation with one hook installed.
3. **Uncertainty is prompt-clustered.** Several seeds of one prompt are not independent
   observations, so every interval resamples *prompts* and carries their seeds along.

## What running it produces

Per question: a tidy table (one row per intervention × layer × head), a paper-ready figure with a
self-contained caption, and a printed verdict. Across questions: a claim-by-claim evidence table
with prompt-clustered confidence intervals, a forest plot, and the causal-circuit diagram with each
arrow coloured by what this checkpoint actually showed.

> **Read it first, run it second.** Section 4 renders every figure from a clearly-labelled mock
> result, so you can see the shape of each answer — and check that the figures say what you want
> them to say — before spending any GPU time.

## 2. The three phases, and why they are separate

| Phase | Purpose | Size | May support a claim in the paper? |
|---|---|---:|---|
| `debug` | Check that hooks, shapes, exports and figures work | 2 prompts × 1 seed | **No** |
| `discovery` | Choose layers, fit the checkpoint's $v^*$, select channels and thresholds | 10 prompts × 5 seeds | Only to *choose* and freeze the protocol |
| `confirmatory` | Test the frozen hypotheses on unseen prompts | 10 prompts × 5 seeds | **Yes** |

If a layer, a channel or a token set is chosen after seeing which intervention gave the nicest
answer, the reported effect is biased upward and a reviewer is right to discount it. Discovery
chooses; confirmation tests. The two phases use **disjoint prompts** and write to different
directories, and `DiscoveryArtifact` refuses to overwrite a frozen manifest.

The full ICLR design calls for a larger confirmatory set (≈200 prompts × 3 seeds). Everything here
scales by editing `PROMPT_COUNT` in §3; the default keeps a first end-to-end run affordable.

```python
# 2.1 — The one cell you edit.
CHECKPOINT = 'tiny-flux1'   # tiny-flux1 | flux1-schnell | flux1-dev | pixart-sigma-512
PHASE      = 'debug'        # debug | discovery | confirmatory

# Staged execution. Nothing runs the full grid by default: the gates below are
# cheap and the experiments are not, and a validated pipeline is worth more than
# an early number.
#   0  gates only (no experiment)        1  1 prompt x 1 seed
#   2  1 prompt x 3 seeds                3  2 prompts x 2 seeds
#   4  the full intended run
STAGE = 0

# DELIBERATELY EMPTY. Populate only after both gates pass and you have set
# PROCEED_TO_EXPERIMENTS = True in the human-checkpoint cell below.
QUESTIONS = []

SEEDS = [0, 1, 2, 3, 4]          # the five pre-declared seeds for every real checkpoint
PROMPT_COUNT = 20                # 10 discovery + 10 confirmatory
FORCE_REDISCOVERY = False        # True only to rebuild a *debug* artifact

# Set to True only by hand, in the human-checkpoint cell, after reading Gate A and
# Gate B output. Declared here so that a "Run all" cannot leave it undefined.
PROCEED_TO_EXPERIMENTS = False

_STAGE_UNITS = {0: (0, 0), 1: (1, 1), 2: (1, 3), 3: (2, 2), 4: (PROMPT_COUNT, len(SEEDS))}
assert PHASE in {'debug', 'discovery', 'confirmatory'}
assert CHECKPOINT in {'tiny-flux1', 'flux1-schnell', 'flux1-dev', 'pixart-sigma-512'}
assert set(QUESTIONS) <= {'q1', 'q2', 'q3', 'q4', 'q5', 'q6'}
assert STAGE in _STAGE_UNITS
_p, _s = _STAGE_UNITS[STAGE]
print(f'{CHECKPOINT} · {PHASE} · stage {STAGE} ({_p} prompt(s) x {_s} seed(s))')
print(f'questions queued: {", ".join(QUESTIONS) if QUESTIONS else "NONE (gates only)"}')
print(f'PROCEED_TO_EXPERIMENTS = {PROCEED_TO_EXPERIMENTS}')
```

```python
# 2.3 — Shared helpers, and the fail-closed guard every experiment cell calls.
import matplotlib.pyplot as plt

from ditsinks import causal_figures as CF          # the six question figures
from ditsinks import causal_stats as CS            # clustered bootstrap, paired effects
from ditsinks import style as ST                   # the paper style
from ditsinks.questions import QUESTION_TITLES, head_level, layer_level

ST.use('paper')                                    # serif type, closed box, CVD-safe palette
RESULTS = globals().get('RESULTS', {})             # keep anything already computed


def require_approval(what='this experiment'):
    """Refuse to run an experiment unless both gates passed and a human approved.

    Defined here, and reading globals defensively, so that it fails CLOSED however
    the notebook is executed: under "Run all", from a half-run kernel, or when a
    cell is run out of order. A missing flag is a refusal, never a pass.
    """
    for flag, why in (('TESTS_PASSED', 'the test suite has not passed in this session'),
                      ('GATE_A_PASSED', 'Gate A (diagnostic harness) has not passed'),
                      ('GATE_B_PASSED', 'Gate B (NeurIPS compatibility) has not passed')):
        if not globals().get(flag, False):
            raise RuntimeError(
                f'REFUSING to run {what}: {why}. Run the validation section above first; '
                'an experiment on unvalidated hooks produces numbers, not evidence.')
    if not globals().get('PROCEED_TO_EXPERIMENTS', False):
        raise RuntimeError(
            f'REFUSING to run {what}: both gates passed, but manual approval is required.\n'
            'Read the Gate A table and the Gate B summary, then set PROCEED_TO_EXPERIMENTS = True '
            'in the human-checkpoint cell and re-run it.')


def staged_units(ctx):
    """The prompt-seed units STAGE allows, so a debug run cannot silently be a full one."""
    prompts, seeds = _STAGE_UNITS[STAGE]
    if not prompts or not seeds:
        raise RuntimeError(f'STAGE {STAGE} runs no experiments. Raise STAGE to 1 or more.')
    keep = {int(s) for s in list(ctx.seeds)[:seeds]}
    units = [u for u in ctx.units() if u[0] < prompts and int(u[2]) in keep]
    if not units:
        raise RuntimeError(f'STAGE {STAGE} selected no units from ctx.')
    return units


def staged_ctx(ctx):
    """A context narrowed to STAGE, so a debug run cannot quietly become a full one.

    The runners read ctx.units(), so bounding the prompts and seeds on the context
    is what actually limits the work -- passing a shorter unit list to one runner
    would not constrain the next.
    """
    import dataclasses
    prompts, seeds = _STAGE_UNITS[STAGE]
    if not prompts or not seeds:
        raise RuntimeError(f'STAGE {STAGE} runs no experiments. Raise STAGE to 1 or more.')
    return dataclasses.replace(ctx, prompts=list(ctx.prompts)[:prompts],
                               seeds=list(ctx.seeds)[:seeds])


print('helpers ready; results held:', sorted(RESULTS) or 'none yet')
print('experiments are guarded by require_approval(); staged_units() bounds their size.')
```

## 3. The prompt manifest

DiffusionDB is a log of real submissions, in submission order. Taking the first N rows therefore
does not sample N users -- it usually samples one person iterating on one idea. A pilot draw of the
first twenty rows returned six variants of *"a renaissance portrait of &lt;celebrity&gt;, art in the style
of rembrandt"*, three of *"portrait of a dancing eagle woman"*, and two each of two more templates:
thirteen of twenty prompts were near-copies of another.

That is not cosmetic. **The prompt is the bootstrap cluster** for every interval in this notebook,
so near-copies inflate the apparent number of independent observations and understate uncertainty;
and when one template lands in both halves of the split, the split stops being a split.

The prompts below are therefore drawn the way the dataset's own documentation recommends -- from
the published `metadata.parquet`, which carries the prompt text together with the submitting
`user_name` and safety scores, and never touches the image archives -- with row groups sampled from
across the file rather than from its start. Four filters then make the set defensible: **at most one
prompt per user**, **content-word overlap below 0.5** between any two kept prompts, the dataset's own
**safety scores**, and a **minimum length**. The exact policy is written into the manifest next to the
prompts.

> Zijie J. Wang, Evan Montoya, David Munechika, Haoyang Yang, Benjamin Hoover and Duen Horng Chau.
> *DiffusionDB: A Large-scale Prompt Gallery Dataset for Text-to-Image Generative Models.*
> ACL 2023, pages 893-911.

```python
# 3.1 — Freeze the prompt split.
import pandas as pd
from ditsinks.prompt_sets import DIFFUSIONDB_CITATION, diffusiondb_prompts, diversity_report

PROMPT_MANIFEST = DRIVE_ROOT / 'protocol' / f'diffusiondb_{PROMPT_COUNT}_prompts.json'
PROMPT_SEED = 0            # changing this draws a different, equally valid set

try:
    ALL_PROMPTS = diffusiondb_prompts(PROMPT_MANIFEST, count=PROMPT_COUNT, seed=PROMPT_SEED)
except ValueError as exc:
    # A manifest written by the earlier order-based draw contains near-duplicates.
    # Keep it for reference rather than deleting it, and draw a fresh set.
    superseded = PROMPT_MANIFEST.with_suffix('.superseded.json')
    PROMPT_MANIFEST.replace(superseded)
    print(f'Superseded the earlier prompt manifest -> {superseded.name}\n  reason: {exc}\n')
    ALL_PROMPTS = diffusiondb_prompts(PROMPT_MANIFEST, count=PROMPT_COUNT, seed=PROMPT_SEED)

half = len(ALL_PROMPTS) // 2
DISCOVERY_PROMPTS, CONFIRMATORY_PROMPTS = ALL_PROMPTS[:half], ALL_PROMPTS[half:]
PROMPTS_BY_PHASE = {'debug': ALL_PROMPTS[:2], 'discovery': DISCOVERY_PROMPTS,
                    'confirmatory': CONFIRMATORY_PROMPTS}
SEEDS_BY_PHASE = {'debug': SEEDS[:1], 'discovery': SEEDS, 'confirmatory': SEEDS}

assert len(set(ALL_PROMPTS)) == len(ALL_PROMPTS), 'prompts must be unique'
assert set(DISCOVERY_PROMPTS).isdisjoint(CONFIRMATORY_PROMPTS)
print(DIFFUSIONDB_CITATION, '\n')
print(json.loads(PROMPT_MANIFEST.read_text())['selection_summary'])
display(pd.DataFrame({'prompt_id': range(len(ALL_PROMPTS)),
                      'split': ['discovery'] * half + ['confirmatory'] * (len(ALL_PROMPTS) - half),
                      'prompt': [p[:96] for p in ALL_PROMPTS]}))

# The diversity audit: the most similar pair in the set, so a reader can see that
# no two prompts are variations on each other.
overlap = diversity_report(ALL_PROMPTS)
print(f'\nmost similar pair in the set: {overlap.overlap.iloc[0]:.2f} content-word overlap')
display(overlap.head(3)[['overlap', 'a', 'b']])
print(f'{PHASE}: {len(PROMPTS_BY_PHASE[PHASE])} prompts x {len(SEEDS_BY_PHASE[PHASE])} seeds '
      f'= {len(PROMPTS_BY_PHASE[PHASE]) * len(SEEDS_BY_PHASE[PHASE])} paired units')
```

```python
# 3.2 — Capture configuration. One denoising step is captured per generation; the register
# lifecycle is a depth phenomenon, so depth is what we sweep, not time.
import json, math, torch
from ditsinks import SweepConfig
from ditsinks.discovery import LayerRanges

MODEL_SETTINGS = {
    # size, steps, and the (writer, register, dissolution) layer ranges frozen from the v5 atlas.
    'tiny-flux1':       dict(size=64,  steps=2,  ranges=LayerRanges((0, 1), (1, 3), (3, 4))),
    'flux1-schnell':    dict(size=512, steps=4,  ranges=LayerRanges((17, 19), (20, 39), (40, 56))),
    'flux1-dev':        dict(size=512, steps=28, ranges=LayerRanges((17, 19), (20, 39), (40, 56))),
    'pixart-sigma-512': dict(size=512, steps=20, ranges=LayerRanges((8, 10), (10, 20), (20, 27))),
}
SETTINGS = MODEL_SETTINGS[CHECKPOINT]
CAPTURE_STEP = SETTINGS['steps'] // 2

cfg = SweepConfig(
    model=CHECKPOINT, prompts=PROMPTS_BY_PHASE[PHASE], seeds=SEEDS_BY_PHASE[PHASE],
    height=SETTINGS['size'], width=SETTINGS['size'], num_inference_steps=SETTINGS['steps'],
    capture_steps=[CAPTURE_STEP], output_dir=str(RUN_ROOT),
    dtype='float32' if CHECKPOINT.startswith('tiny') else 'bfloat16', save_images=True)
print(cfg.to_json())
# save_images=True is what puts the generated pictures on disk: the discovery
# sweep writes one per prompt-seed under its own folder, and section 7 saves the
# image its maps are laid over beside the maps themselves. They are small next to
# the activations and they are the one output a reader looks at directly.
```

## 5. Clean discovery, then freeze

## 6. The instrumentation check that everything else depends on

Before believing any causal result, confirm that *installing* the hook changes nothing. A sham hook
reads the image tokens, clones them and writes them straight back. If the resulting trajectory is
not identical to the clean one, the instrumentation itself is perturbing the model and no condition
downstream can be interpreted.

This is a hard assertion, not a diagnostic. The engine keeps hooks out of the numerics on purpose:
the attention tap calls the model's own kernel and computes statistics in `float32` alongside it,
and an edit hook writes back a tensor of identical shape and dtype.

```python
# 6.1 — A sham hook must reproduce the clean trajectory exactly.
from ditsinks.adapters import InterventionPoint
from ditsinks.causal_engine import EditPlan, run_traced_generation, select_frozen_targets

prompt_id, prompt, seed = ctx.units()[0]
tracer = ctx.tracer()
clean_trace, _ = run_traced_generation(
    driver, tracer, prompt_id=prompt_id, prompt=prompt, seed=seed,
    probes=[(InterventionPoint.BLOCK_INPUT, [ctx.intervention_layer])])
frozen = select_frozen_targets(clean_trace, layer=ctx.intervention_layer, step=ctx.step,
                               point=InterventionPoint.BLOCK_INPUT, direction=ctx.vstar)
sham_trace, sham_stats = run_traced_generation(
    driver, tracer, prompt_id=prompt_id, prompt=prompt, seed=seed, condition='sham',
    plans=[EditPlan(edit=lambda image, _: image.clone(), point=InterventionPoint.BLOCK_INPUT,
                    layers=[ctx.intervention_layer], label='sham')], targets=frozen)

rows = []
for layer in ctx.downstream_layers:
    a, b = clean_trace.at(ctx.step, layer), sham_trace.at(ctx.step, layer)
    if a is None or b is None or a.projection is None:
        continue
    rows.append({'layer': layer,
                 'max |change| in direction projection': float((a.projection - b.projection).abs().max()),
                 'max |change| in token norm': float((a.norm - b.norm).abs().max())})
check = pd.DataFrame(rows)
display(check)
print(f'sham hook fired {sham_stats.edit_calls} times; '
      f'frozen registers: {frozen.register_ids}; matched ordinary: {frozen.matched_ordinary_ids}')
worst = float(check.iloc[:, 1:].to_numpy().max()) if len(check) else 0.0
assert worst == 0.0, (f'The sham hook changed the trajectory by up to {worst:.3e}. '
                      'Stop: no causal result from these hooks is interpretable.')
print('PASS — installing the hook changes nothing.')
```

## 6.5 Validation gates — nothing below runs until both pass

The repaired pipeline has never been executed against real model tensors. Codex could
repair it but had no PyTorch runtime, so every empirical gate in
`docs/iclr_causal_validation_status.md` is recorded as **NOT RUN**. These two cells are
what close that gap.

**Gate A — `run_diagnostic_harness(ctx)`.** One prompt, one seed. It asserts, on live
tensors, that every manipulation does what it claims and every identity holds: the norm
clamp preserves cosine and hits its target norm; direction removal leaves no projection;
the dominant channel is *exactly* zero; a sham hook reproduces states, keys, queries,
attention logits, attention probabilities, sink ids and the final image bit-for-bit; and
the final-key and full-residual self-patches are identities. It raises `AssertionError`

**Gate B — `run_neurips_compatibility(ctx)`.** Reproduces the established NeurIPS

separation is the one the workshop paper established:

| Intervention | Established expectation |
|---|---|
| reduce norm, keep direction | most **affected** heads KEEP the natural sink (historically ~89%) |
| change direction, keep norm | most **affected** heads ABANDON it (historically ~4%) |

**Read this before running Gate B.** Unlike Gate A, `run_neurips_compatibility` does
**not** assert anything — it returns a summary table and a verdict string saying

## 7. What each structure's removal does to the others

Before the six questions, one picture of the thing they are all about.

The workshop paper's claim is that high-norm tokens, the dominant activation channel and the
attention sinks are **one** phenomenon rather than three that happen to coincide. A correlation
cannot settle that. Taking each one away and looking at the other two on the same image patches can.

The three removals deliberately act at different places in the model, so that none of them removes
another by construction:

| Removal | Where it acts | What it leaves alone |
|---|---|---|
| High-norm tokens | replaces the frozen register states with matched ordinary states in the residual stream | everything downstream is free to rebuild them |
| Dominant channel | scales that channel to zero in the writing block's contribution, and holds it there through the register zone | every other channel of the same tokens |
| Attention sinks | replaces only the **keys** those positions present to attention, with the mean key of the ordinary patches | the residual state entirely — norm, channel and direction are untouched |

The third is the one that separates **representation** from **routing**. If those tokens keep their
norm and their channel while their sink is gone, then anything that changes downstream is caused by
where attention went, not by the state that attracted it. Replacing the key with the ordinary mean
rather than zeroing it matters: a zero key still scores zero against every query, which can be the
largest score when the others are negative, and the sink would survive the intervention meant to
remove it.

### Two methods, and why both are here

Every panel below is filled one of two ways, and the figures say which on every panel because they
answer different questions.

* **Recomputed** is bookkeeping. The same quantity is recomputed from the untouched states with the
  structure masked out of the arithmetic, and the model is never re-run. It answers *how much of
  what we are looking at was only this structure being counted*, and it is exact — nothing
  downstream had a chance to react, so there is nothing to confound it. It also costs no extra
  generation.
* **Model re-run** generates again with the structure taken away, so every block after the
  intervention is free to respond. It answers *what does the network do about it*, which is the
  question the paper is asking, at the price of every downstream reaction being part of the answer.

Which one a row can use is forced rather than chosen. Bookkeeping exists only where the measured
quantity is a function of the captured residual-stream states **and** the removal is a mask on those
same states. Attention is neither: it is formed inside the block from queries and keys, so masking a
channel out of the residual stream afterwards says nothing about where a head would have looked, and
key-space sink suppression is not a mask on the residual stream at all. Those panels can only be
filled by running the model.

**How to read the figures.** 7.2 is the strip: one row per question of the form *take this away,
then look at that*, laid on the latent patch grid, each row carrying its own evidence — the image,
the structure that was taken away, the measured structure before, and the measured structure after.
Circles mark the frozen register patches; the badge reports how much of the measured structure is
left there. 7.3 is the same nine measurements compressed into one grid of panels, for reading the
whole pattern at once, and 7.4 is the same numbers as a table. Colour scales are fixed from the
untouched run, so a panel going dark means the structure went away rather than that the scale moved
with it. Rows and cells outlined in red are the manipulation check, where a removal is measured
against itself: they confirm each intervention did what it claims, and are not findings.

## Q1. If the natural register tokens are removed, where do the attention sinks go?

We know the high-norm registers usually *are* the sinks, and that direction rather than magnitude
decides which token a head sinks on. What nobody has asked is what the network does when the
register population itself is taken away. This is the strongest available test of whether the
observed co-localisation is causal, and it is the experiment that decides whether register-aware
token pruning is worth pursuing at all: if removing registers merely triggers cheap relocation,
protecting them during pruning buys nothing.

**Where we intervene.** At the input of the first clear register layer — the state the writer has
just produced. Targets are the clean top-1 % high-norm image tokens, plus a threshold-free top-$k$
companion so no headline claim rests on one percentile.

**Four interventions, each isolating a different property of the register.**

| Condition | What it removes | What it keeps |
|---|---|---|
| Register direction removed | the $v^*$ component | magnitude, and every other channel |
| Replaced by matched ordinary state | everything | nothing — but the replacement is a realistic ordinary state |
| Clamped to ordinary norm | magnitude | direction |
| Register state zeroed | everything | nothing; the deliberately extreme perturbation |

**Five controls, so effect size cannot be attributed to perturbation size.** A sham hook; the same
zeroing applied to count-matched random tokens; the same zeroing applied to the highest-norm tokens
that are *not* registers; an equal-energy direction removal applied off-register; and -- matched in
*strength* to the headline intervention rather than to zero -- the identical direction removal
applied to those same norm-matched non-registers. Zeroing a token is far more violent than removing
its direction, so without that last control a difference between the two could be about how hard we
perturbed rather than which tokens we chose.

**What each outcome would mean.**

* *Attention spreads out* → natural registers are indispensable routing anchors.
* *A reserve token takes over* → the model keeps a ranked reserve of candidate registers.
* *The same positions regenerate* → a persistent writer is tied to those positions.
* *New positions acquire $v^*$* → the model preserves a register **function**, not a fixed spatial identity.

**Pre-declared endpoints.** Primary: the fraction of *affected* heads (those whose clean sink was a
removed token) still sinking on the clean token, and the change in the targets' geometry.

Geometry is reported as **two** numbers, not one. The projection on $v^*$ is $\lVert x \rVert \cos\theta$
— the product of magnitude and alignment — so a single projection change cannot tell "the token
stopped pointing along $v^*$" from "the token got smaller", which is exactly the distinction the
whole account rests on. Panel (d) plots them on separate axes, and each condition lands in the
quadrant that says what it actually did.

The direction removal is **norm-preserving** for the same reason: $x - \langle x, v^*\rangle v^*$ has
norm $\lVert x \rVert\sqrt{1-\cos^2\theta}$, so for a register nearly collinear with $v^*$ a plain
removal would leave it both directionless and tiny. Everything else — attention concentration,
entropy, key rank, query-key advantage, takeover channel — is a diagnostic.

## Q2. What happens to $v^*$, key geometry and sink identity after live dominant-channel intervention?

That channel 154 accounts arithmetically for the high-norm signature is settled. The open question
is what happens when it is suppressed *during* the forward pass, so that every later block sees the
suppressed state. That is the difference between a numerical explanation and a causal one.

**Where we intervene.** At the writer layer's residual contribution — the projected update *before*
it is added to the residual stream — at the frozen register positions.

**Two scopes.** `writer_only` scales the channel once, at the writing block. `maintenance` keeps
scaling it through the following register-zone blocks. If the two agree, the register is written
once; if maintenance is much stronger, later blocks are actively rebuilding the state.

**A dose, not a point.** $\gamma \in \{0, 0.1, 0.25, 0.5, 0.75, 1.0, 1.5\}$. $\gamma = 1$ is the
sham scale — the hook fires and multiplies by one — which makes it the exact reference for the
paired contrast. $\gamma = 1.5$ tests amplification, which a purely destructive ablation cannot.

**Five controls at full suppression.** The late-growing competitor; a massive but
register-unspecific channel; a random channel; an energy-matched removal along a direction
orthogonal to $v^*$; and the same scalar applied at ordinary positions. Together they separate
"this channel" from "a channel", "this much energy", and "these positions".

**What each outcome would mean.** Suppression lowers $v^*$ *and* sink behaviour → the channel is a
proximal causal carrier of the routing state. High norm disappears but $v^*$ and the sink recover →
downstream computation rewrites the register. Another channel becomes dominant and preserves the
sink → the circuit has redundancy.

## Q3. Does the network regenerate or relocate $v^*$ after we destroy it?

Direction ablation is already known to cost the sink immediately. What it does not tell us is what
the *next* blocks do. If the direction comes back, the register is not a one-time write but an
actively maintained state — and that changes the mechanism, and the application, completely.

**The intervention.** Destroy the $v^*$ component at a natural register while preserving the token's
original magnitude, so a token that recovers cannot have recovered merely by shrinking.

**Two schedules.** `single_shot` removes once at the register layer. `repeated` removes again at
every register-zone block. If a single shot recovers but repeated removal does not, later blocks are
doing the rebuilding.

**Restoration, to localise the pressure.** Additional conditions patch the *clean* residual back at
one later layer. The layer at which that restores downstream behaviour is where maintenance pressure
is strongest.

**The readout that matters.** Not just whether the sink returns, but **where**: at the same tokens
(a writer bound to those positions), at new tokens (a register function without a fixed spatial
identity), or not at all (a one-time write). We track the strongest alignment with $v^*$ over *all*
image tokens at every downstream block, and record the first layer that crosses the frozen clean
threshold.

## Q4. What selects the sparse token positions that receive the register write?

We know *what* becomes extreme. The mechanistic question is why only a handful of image tokens are
chosen. Answering it completes the beginning of the circuit — and it is the one place where a purely
observational answer is worthless, because anything that correlates with becoming a register also
correlates with everything else that register does.

**Backward prediction, as a localisation tool only.** At the writer layer we read four candidate
pre-write features and ask how well each separates eventual registers from matched ordinary tokens,
using a rank-based separability score where 0.5 is chance:

* the pre-feed-forward residual,
* the feed-forward hidden activation,
* the timestep-modulated normalised stream,
* the token's pre-existing $v^*$ component.

**Reciprocal patching, which is the actual claim.** For each feature we copy it from an eventual
register into a matched ordinary token (*transfer*), and separately replace it at an eventual
register with the matched ordinary value (*prevention*). A feature only counts as a writer when
**both** directions move as predicted; the figure draws the two as opposing bars so a one-sided
feature is visible immediately.

**Endpoints.** Does the recipient develop the dominant channel? A high norm? Alignment with $v^*$?
And later, does it become a sink? Each is paired against a sham hook on the same prompt, seed and
noise.

**A note on architecture honesty.** Intervention points resolve through capability checks. In FLUX
the adaptive-norm gates are computed per timestep and shared across tokens, so they cannot select
*which* token becomes a register; the notebook reports that as a measured property (the
`token_dependence` column) rather than assuming it. Any stage the architecture does not expose as a
separable module is reported unsupported and left untested — never approximated with a neighbouring
tensor.

## Q6. What causes the register state to dissolve?

The lifecycle is described; the cause of its end is not. The strongest hypothesis is competition: a
second channel grows through the late blocks until $v^*$ is no longer dominant. In FLUX the
candidate is channel 1446; in PixArt-Σ it is 295 against 293. The alternative is that the state
simply rotates away, which no single channel would explain.

**The decomposition.** Through the late register zone we write each register state as
$x = \alpha v^* + x_\perp$ and track $\alpha$, $\lVert x_\perp \rVert$, the dominant channel's
contribution, the competitor's contribution, angular velocity away from $v^*$, key rank and sink
strength.

**Three interventions and three controls.**

* *Competing channel suppressed late* — if the competitor causes dissolution, removing it should
  extend the register's life.
* *Competing channel amplified early* — the converse; dissolution should arrive sooner.
* *Register direction refreshed at unchanged norm* — re-project the state towards $v^*$ at each late
  block **without** changing its magnitude. This separates rotation from magnitude decay.
* Controls: a sham hook, a random channel suppressed on the same schedule, and an energy-matched
  removal along a direction orthogonal to $v^*$.

**The endpoint.** Register lifetime: the last observed layer at which the *frozen* register tokens
are still both high-norm and aligned — frozen identities, so an intervention cannot buy lifetime by
promoting a different token. When no layer crosses that threshold, the notebook falls back to the
alignment half-life, a continuous companion that is always defined, and says which one it used.

**How to read the result.** Channel suppression prolongs the register → channel competition. Only
the direction refresh prolongs it → dissolution is a geometric rotation that one channel does not
explain. Neither → the model is erasing or re-encoding the register through a broader subspace, and
the single-channel hypothesis is wrong.

## 9. Statistics, exports and stopping rules

**The unit is the prompt-seed pair; the cluster is the prompt.** Effects are within-unit
differences against a matched reference run, and every interval resamples prompts with replacement,
carrying their seeds along. Pairing removes prompt-to-prompt variance from the effect; clustering
stops the interval from pretending five seeds are five independent observations.

**Primary endpoints are pre-declared**, two per question at most: the fraction of affected heads
retaining the clean sink, and the change in the register direction's projection. Everything else is
a mechanistic diagnostic and is reported as such.

**Near-null claims get an equivalence test**, not a failed significance test. The margin is declared
in `ditsinks.causal_stats.PRIMARY_CLAIMS` before the run.

**Stopping rules.** Stop and fix rather than report if: the sham hook in §6 changes the trajectory;
discovery fails to reproduce the expected channel *and* you were treating the channel as known; or a
question's primary endpoint is undefined because no head ever sank on a frozen target — in that
last case the verdict says so, and the answer is a bigger clean run, not a different endpoint.

The final cell records everything needed to reproduce this run: the config, the discovery manifest,
the git commit, package versions and the frozen prompt split.

```python
# 9.1 — Equivalence tests for the claims that are supposed to come out null.
null_claims = [c for c in CS.PRIMARY_CLAIMS if c.expect == 'equivalent']
rows = []
for claim in null_claims:
    result = RESULTS.get(claim.question)
    if result is None:
        continue
    frame = result.tidy if claim.table == 'tidy' else result.tables.get(claim.table)
    if frame is None or frame.empty:
        continue
    if claim.filter_layer_level and 'head' in frame.columns:
        frame = layer_level(frame)
    differences = CS.paired_differences(frame, claim.endpoint, condition=claim.condition_column,
                                        treatment=claim.treatment, reference=claim.reference)
    outcome = CS.equivalence(differences, claim.endpoint, margin=claim.margin,
                             label=claim.name, iterations=2000)
    rows.append({'claim': claim.name, 'margin': outcome.margin, 'effect': outcome.effect,
                 '95% CI': f'[{outcome.ci_low:.3f}, {outcome.ci_high:.3f}]',
                 'practically equivalent': outcome.equivalent, 'units': outcome.n_units})
display(pd.DataFrame(rows).round(3) if rows else
        pd.DataFrame(columns=['claim', 'margin', 'effect', '95% CI', 'practically equivalent']))
```

## 10. Recommended execution order and what each pass costs

> **The gates come first, always.** Repository setup → tests → model/discovery/ctx →
> Gate A → Gate B → human approval → experiments. Nothing below overrides that order;
> `require_approval()` refuses every experiment cell until all of it has happened in the
> current session.

**Stages.** `STAGE` in cell 2.1 bounds how much any experiment runs, so a debug pass cannot
quietly become a full one:

| STAGE | Units | Use |
|---|---|---|
| 0 | none | gates only — the default |
| 1 | 1 prompt × 1 seed | first validated experiment |
| 2 | 1 prompt × 3 seeds | seed sensitivity |
| 3 | 2 prompts × 2 seeds | prompt sensitivity |
| 4 | full | only once 1–3 look right |

**Gate B needs a real checkpoint.** On `tiny-flux1` no head sinks on a frozen register in the
clean run, so the affected-head denominator is empty and Gate B reports zero affected heads
and hard-stops. That is correct behaviour, not a failure: use `tiny-flux1` to exercise
plumbing and Gate A, and `flux1-schnell` to actually clear Gate B.

1. **`tiny-flux1`, `debug`.** Runs on CPU with synthetic weights in a couple of minutes. It exercises
   every hook, every export and every figure. It cannot support a scientific claim — the tiny model
   has no real register circuit — but if it fails, nothing larger will work.
2. **`flux1-schnell`, `discovery`.** Four denoising steps makes it the cheapest real checkpoint.
   Inspect the v5 atlases, accept the layer ranges, and freeze the artifact. Run Q1, Q2, Q3 and Q6
   first: they are the P0 core.
3. **`pixart-sigma-512`, `discovery` then the compact chain.** Cross-architecture causal replication
   needs one decisive chain — dominant-channel suppression → loss of $v^*$ and sink → selective
   restoration — not a duplicate of every FLUX experiment. PixArt also exposes the pre-position key
   that FLUX does not, so Q5 gains a rung there.
4. **`flux1-dev`, `confirmatory`.** Only after the hooks and controls pass elsewhere. This is the
   expensive one.
5. **Q5 then Q4.** Their hook locations are the most architecture-dependent; run them once the
   cheaper questions have shown the instrumentation is sound.

Each question runs one clean pass plus one generation per condition per prompt-seed unit, so cost
scales as (conditions + 1) × prompts × seeds. To shorten a first real pass, cut `Q2_GAMMAS` to
`[0.0, 0.5, 1.0]` and run `QUESTIONS = ['q1', 'q2']`.

**A disconnect is safe** only after a question has written its tables: each `run_qN` saves to
`RUN_ROOT/<question>/` as it completes, and the discovery artifact is never rewritten. Do not pool
checkpoints — each has its own $v^*$, its own channels and its own manifest.
